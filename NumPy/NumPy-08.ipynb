`Broadcasting`

```python
# concept
```

```python
start = np.zeros((4, 3))
start
```

```python
# create a 1d ndarray with 3 values
add_rows = np.array([1, 0, 2])
add_rows
```

```python
print(start.shape, add_rows.shape)
```

```python
# add to each row of 'start' using broadcasting
y = start + add_rows
y
```

```python
# create an ndarray which is 4 x 1 to broadcast across column
add_cols = np.array([[0, 1, 2, 3]])
add_cols = add_cols.T
print(add_cols, add_cols.shape)
```

```python
# add to each column of 'start' using broadcasting
start + add_cols
```

```python
# this will just broadcast in both dimensions
add_scalar = np.array([100])
add_scalar.shape
```

```python
print(start + add_scalar)
```

```python
# here boardcasting is impossible
A = np.zeros((2, 3))
b = np.array([1, 2])

A + b
```

```python
mat1
```

```python
# Dividing each value of mat2 by mat1
mat2 / mat1
```
