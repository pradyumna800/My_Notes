## NumPy-07: Indexing, Slicing and Subsetting in NumPy Arrays

`1. Indexing`
* `Definition`: Accessing idivisual elements of a NumPy array using their posititon (index).
* `Key Point`: Indexing starts at 0 in Python

```python
import numpy as np
arr = np.array([10, 20, 30, 40])
print(arr[2]) # Output: 30
```

`2. Slicing`
* `Definition`: Extracting a `range of elements` from a NumPy array using the [start:stop:step] notation.
* `Key Points`: Slicing create a `view`, not a copy in NumPy (Means if i do any changes on sliced array then it also reflect on original array).

```python
arr = np.array([10, 20, 30, 40, 50])
print(arr[1:4]) # Output : [20 30 40]
print(arr[::2]) # Output : [10 30 50]
```

`3. Subsetting`
* `Definition`: Selecting a `portion of the array` using `Boolean condition` or `fancy indexing`.
* `Key Point`: Subsetting allows you to extract elements based on `conditions` or `specific indices`.
* It create a copy - if we do changes on new array then it will not afect on original array. 

```python
arr = np.array([10, 20, 30, 40, 50])
print(arr[arr > 25]) # Output: [30, 40, 50]
```

```python
arr > 25
```

```python
print(arr[[0, 2, 4]]) # Output : [10 30 50]  -> it will create a copy 
```

`Slicing`

```python
mat = np.array([[11, 12, 13], [21, 22, 23], [31, 32, 33]])
mat
```

```python
# Slicing and changing the sliced matrix
mat_slice = mat[:2, :2]
mat_slice
```

```python
mat_slice[0, 0] = 111
```

```python
mat_slice
```

```python
#But the original matrix?
mat  # update happend
```

```python
# Little different way to create a copy of the sliced matrix
mat = np.array([[11, 12, 13], [21, 22, 23], [31, 32, 33]])
mat
```

```python
# Notice the np.array command to create a new array not just slicing
mat_slice = np.array(mat[:2, :2])
mat_slice
```

```python
# Change the sliced matrix
mat_slice[0, 0] = 111
```

```python
mat_slice # changed
```

```python
# But the original matrix ?
mat  # update does not happend
```
