### Variance and Standard Deviation

| Concept | Formula | Description / Notation |
|---|---|---|
| **Variance** | \(\displaystyle \sigma^2 = \frac{\sum (x-\bar{x})^2}{n}\) | \(x\) = Observations given<br>\(\bar{x}\) = Mean<br>\(n\) = Total number of observations |
| **Standard Deviation** | \(\displaystyle S = \sigma = \sqrt{\frac{\sum (x-\bar{x})^2}{n}}\) | \(x\) = Observations given<br>\(\bar{x}\) = Mean<br>\(n\) = Total number of observations |

```python
# Standard deviation of all number in the modified matrix, a larger number
np.std(mat1)
```

```python
# variance of all number in the modified matrix, a larger number
np.var(mat1)
```

## Dot Product

Dot-Product Generalizes inner product to vectors and matrix multiplication for higher dimensions.

- **1-D dot product or inner product** - Sum of product of elements.
- **2-D dot product** is equivalent to matrix multiplication - can be done using `@` as well.

### Comparison of Operations

| Operation | Input Types | Output Type | Formula / Meaning |
|---|---|---|---|
| **Inner Product** | 1D vectors | Scalar | Sum of elementwise products |
| **Outer Product** | 1D vectors | Matrix | All pairwise products |
| **Dot Product** | Vectors or matrices | Scalar or matrix | Vector inner product or matrix multiplication |

```python
A = np.arange(1, 6)
B = ri(1, 10, 5)
```

```python
print(A)
print(B)
```

```python
np.dot(A, B)
```

```python
np.inner(A, B) # Inner Product of vectors A and B (sum of all pairwise elements)
```

```python
# Outer product of vexctors A and B
np.outer(A, B)
```
