`1. ravel()`
* ravel() is NumPy method used to flatten a multi-diamentional array into a 1D array.
* It returns a views of the original array whenever possible (i.e. it does not copy data unless necessary)

```python
# help
help(np.ravel)
```

```python
a
```

```python
np.ravel(a) # row major order
```

```python
# Row-major (C-style, last axis changes fastest)
np.ravel(a, order='C')
```

```python
# Column-major (Fortran-style, first axis changes fastest)
np.ravel(a, order='F')
```

```python
# reshape
a.reshape(-1)
```

```python
a.reshape(2, 3)
```

```python
a.reshape(2, -1)
```

```python
# flatten()
a.flatten()
```

```python
temp = np.ravel(a)
temp
```

```python
# reshape
np.reshape(temp, (3, 2)) # row major
```

```python
# Fortram-like index ordering => Column-major
np.reshape(temp, (3, 2), order = 'F')
```

## Difference between `ravel()` and `flatten()`

- `ravel()` flattens the array to 1D.
- Returns a **view** if possible, else returns a copy.
- Faster than `flatten()` because it avoids copying when possible.

### Comparison: `ravel()` vs `flatten()`

| Feature | `ravel()` | `flatten()` |
|---|---|---|
| **Returns** | View (if possible) | Always returns a copy |
| **Memory Efficient** | Yes | ❌ No |
| **Modifying result affects original?** | Yes (if view) | No (copy is separate) |

## View

- When you create a view (via slicing, reshape, transpose, etc.):
- NumPy does **NOT** copy the data.
- Instead, it creates a new ndarray object that has separate (shape, strides, etc.) metadata but points to the same data buffer.

```python
import numpy as np

arr = np.array([10, 20, 30, 40, 50, 60])
view = arr[1:5:2] # select [20, 40]
```

```python
view
```

```python
view[0] = 123
```

```python
arr  # original array changes
```

```python
# to check that two var are share the same memory or not
np.shares_memory(arr, view) # True -> it's a view
```

## 2. `moveaxis()`

- `np.moveaxis()` is used to reorder axes in a NumPy array by moving specific axes to new positions, while keeping the order of the other axes the same.
- Returns a **view** of the array with moved axes.

```python
x = np.zeros((3, 4, 5))
print(x)
```

```python
print(x.shape)
```

```python
np.moveaxis(x, 0, -1).shape
```

```python
print(x.shape)
np.moveaxis(x, [0, 1], [-1, -2]).shape
```

## `4. concatenate()`
* In NumPy, concatenate() is used to join two or more arrays along an axix.

```python
# Join a sequence of array along an existing axis
```

```python
a = np.array([[1, 2], [3, 4]])
```

```python
b = np.array([[5, 6]])
```

```python
a.shape
```

```python
b.shape
```

```python
# default axis=0
np.concatenate((a, b ))
```

```python
np.concatenate((a, b )).shape
```

```python
# dimension will increse along axis 0
np.concatenate((a, b ), axis=0)
```

```python
# dimension will increse along axis 1
np.concatenate((a, b.T ), axis=1)
```

```python
np.concatenate((a, b.T ), axis=1).shape
```

## `5. stack()`
* In NumPy, stack() is used to join a sequence of arrays along a new axis -- unlike concatenate() which joins along an existing axis

```python
a = np.array([[1, 2, 3]])

b = np.array([[2, 3, 4]])
```

```python
a.shape
```

```python
b.shape
```

```python
# default axis
np.stack((a, b))
```

```python
np.stack((a, b)).shape
```

```python
# axis = 0
np.stack((a, b), axis=0)
```

```python
np.stack((a, b)).shape
```

```python
# axis = 1
np.stack((a, b), axis=1)
```

```python
np.stack((a, b), axis=1).shape
```

```python
# hstack()
np.hstack((a, b))
```

```python
# vstack()
np.vstack((a, b))
```

```python
# dstack()
np.dstack((a, b))
```
