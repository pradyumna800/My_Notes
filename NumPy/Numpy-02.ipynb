#### record array or structured array

* Structured arrays in numpy are homogeneous at the array level (each element has the same layout)
* We can specify a type and, optionally, a name on a pre-column basis. This makes sorting and filtering even more powerful, and it can feel similar to working with data in Excel, CSVs, or relational databases.

```python
record = [
    ('aman', 22, 90),
    ('ankit', 15, 60),
    ('akash', 18, 90),
    ('aditya', 22, 80)
]
```

```python
record
```

```python
data = np.array(record, dtype=[('name', 'U15'), ('age', np.int32), ('marks', np.int32)])
data
```

##### You can specify the dtype (data type) in several equivalent ways.
1. Using NumPy type strings (Compat format) - [('name', 'U15'), ('age', np.int32), ('marks', np.int32)]
2. Using NumPy type object (recommended) - [('name', np.unicode_), ('age', np.int32), ('marks', np.int32)]
3. Mixing Python built-in types and NumPy - [('name', str), ('age', int), ('marks', np.int32)]

```python
data[0]
```

```python
data['name']
```

#### By default, NumPy arrays (ndarray) are homogeneous, meaning all elements must be of the same data type.
* Homogeneous arrays allow:
	1. Vectorized operations
	2. Efficient memory layout (contiguous blocks)
	3. Fast computations using C under the hood
* NumPy is designed for arrays of fixed-size elements — typically numerical data types (integers, floats), but also supports strings, booleans, dates, and custom structured types.
* All elements in a NumPy array share the same data type (dtype).
* The data is stored in a contiguous block of memory for efficiency.

* It can support heterogeneous types using object array or structured dtype, but these lose most performance benifits.

```python
het = np.array([5, 4.5, True, 6.5, 'c'])
het
```

```python
het.dtype
```

```python
het[0]
```

```python
het = np.array([5, 4.5, True, 6.5, 'c'], dtype=object)
het
```

```python
het.dtype
```

```python
print(type(het[0]))
```
